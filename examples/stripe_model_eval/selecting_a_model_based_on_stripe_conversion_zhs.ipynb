# 基于 Stripe 转化率选择模型：面向初创公司的实用评估方法
## 概述
最适合你的模型取决于你的业务目标。许多初创公司根据离线评估和公开基准来选择大语言模型（LLM）。然而，在基准测试中得分很高的模型，并不一定能让你的用户付费、订阅或持续使用你的产品。纸面上看起来很强的模型，在用实际业务成果来衡量时可能表现不佳。

本指南介绍一种以初创公司最重要的业务成果之一为依据的评估方法：用户是否愿意为你的产品付费。 

我们将详细介绍 HyperWrite 的模型评估流程，重点关注真实的付费转化，具体来说是一次性购买或月度经常性收入（MRR）订阅的 Stripe 付款。如果你的目标是提高转化率，或者在切换到更便宜的模型时保持转化率，那么这个评估示例或许是一个值得借鉴的模式。
## 前提条件与适用范围
要将本指南应用于你的业务，你需要：

- **一个支付处理商。** 本示例使用 Stripe，但你只需稍作调整，即可将同样的方法用于任何支付服务提供商。
- **足够多的用户以产生有意义的信号。** 每个测试变体的目标用户数至少为一千。要获得更高的统计显著性，你需要更多用户。
- **一个具有转化事件的 AI 驱动产品。** 我们使用的是一个 LLM 应用，转化事件是付费。同样的测试方法也适用于围绕语音、视频和其他模态构建的应用。
## 根据实际目标选择模型
HyperWrite 构建 AI 驱动的写作工具和研究助手。该公司的核心产品是一款具备高级研究能力的写作助手。

离线基准无法预测对 HyperWrite 最重要的事情：用户与写作助手的互动方式是否会促使他们订阅并持续使用该产品。HyperWrite 团队转而关注真正关心的结果，也就是转化，并开始基于比较 Stripe 转化率的真实 A/B 测试在 AI 模型之间做出选择。
## 对初创公司真正重要的是：转化
在许多初创公司，让用户注册并持续使用产品就是目标。借助经典的 A/B 测试，采用科学家们几十年来一直依赖的统计方法，你就可以设计一套模型评估流程：
- 将新用户分批，每批使用不同的 AI 模型。
- 为了统一用户遇到升级提示的时机，在用户向助手发送一定数量的消息后施加一致的速率限制，消息数量足以形成一个有意义的升级时刻。
- 跟踪每组用户向付费订阅（通过 Stripe）的转化情况。

将用户随机分配到不同模型并控制其他因素（新用户引导、功能、提示词等），就可以把转化率的差异归因于被测试的模型，而不是外部变化。统计方法则让我们有把握认为观察到的差异不太可能是偶然造成的。

当发现真实的、非随机的改进时（例如某个模型带来了更高的转化率），影响是实实在在的：更高的 Stripe 转化率、更多的付费用户，而且如果模型更高效，成本往往也更低。
## 如何通过 A/B 测试选择模型
A/B 测试可以作为模型选择的真实世界评估工具。将用户随机分组，为每组提供不同的体验（这里是不同的 AI 模型），然后观察哪一组在关键指标上表现更好，本例中的关键指标就是 Stripe 转化。
### 基础：一个模型对比另一个模型
标准设置包括一个“对照组”（你当前的模型）和一个“变体”（挑战者）。用户被随机分配到其中一组。为了确保测试只隔离模型的影响，其他一切都保持不变：新用户引导、功能、提示词以及转化机会。在达到预先确定的时长或用户数量后，比较转化率：使用模型 A 还是模型 B 时付费的人更多？
### 真实案例：HyperWrite 的模型替换测试
HyperWrite 的目标是部署一个更便宜的 LLM，同时不显著降低变现能力。这是一个非劣效性场景：关注点在于确保新模型不会明显差于对照组。出于节省成本的考虑，团队设计了一个单侧非劣效性检验。
- **测试重点：** 在不损害 Stripe 转化的前提下节省成本。
- **设计：** 单尾双比例 Z 检验（重点检测新模型是否更差）。
- **Alpha（第一类错误率）：** 0.15（即 85% 置信度）。对于这家初创公司来说，迭代速度优先于非常严格的显著性阈值。
- **检验效能：** 0.60（足以发现有意义的下降，同时兼顾流量限制）。
- **最小可检测效应（MDE）：** 转化率下降 30%。如果节省的成本足够合理，任何小于此幅度的下降都会被视为“足够接近”。
- **样本总体：** 某一特定时期内的一部分新注册用户，在注册时按 `user_id` 随机分组。
- **触发条件：** 用户发送消息，遇到升级付费墙，并可能通过 Stripe 结账完成转化。
## 设定参数：怎样才算胜出？
并非每个观察到的差异都有意义，有些差异是偶然产生的。A/B 测试有助于将真实效应与随机噪声区分开。这里常用的统计工具是“双比例 Z 检验”，它检查两组之间的转化率差异是否大到足以被视为具有统计显著性。

这种检验有几种变体：
- **单尾检验：** 检查新模型是否优于对照组（或者根据设计，检查是否不差于对照组）
- **双尾检验：** 检查是否存在任何差异，无论是上升还是下降
- **多变量检验（A/B/n）：** 同时比较三个或更多模型

选择哪种取决于你的目标。如果你要求转化率有明确提升，那么寻找改进的单尾检验可能就足够了。如果你愿意采用一个不差但更便宜的模型，可以设计一个非劣效性（单侧）检验，以确保新模型不会明显更差。
### 关键术语
- **第一类错误（假阳性）：** 在实际不存在效应时得出存在效应的结论
- **第二类错误（假阴性）：** 未能检测到真实存在的效应
- **Alpha (α)：** 可接受的第一类错误风险（通常设为 0.05，即 5%）
- **检验效能：** 检测到真实效应的概率（常见目标为 80%）
### 示例：运行一次真实的模型测试
假设你要在当前模型（对照组）和一个新变体（模型 X）之间做出选择。假设你运行单尾双比例 Z 检验，看模型 X 的转化是否优于对照组。你设定 α = 0.05，并根据基线转化率和期望的最小可检测效应进行检验效能计算后，确定每组大约 1,500 名用户可以提供约 75% 的检验效能，这是为了更快获得方向性洞察而做出的折中。

两组都达到所需样本量后，数据可能如下所示：

| 组别                      | 分配用户数 | 转化数 | 转化率 | p 值 | 统计显著？ | 胜出？ | 已控制第一类错误？ | 已控制第二类错误？ |
|----------------------------|----------------|-------------|-----------------|---------|--------------------|---------|-----------------------|------------------------|
| 对照组（当前模型）    | 1500           | 15          | 1.0%            | --      | 参照          | 否      | 是                   | 是                    |
| 模型 X（变体）          | 1500           | 30          | 2.0%            | 0.012   | 是                | 是     | 是                   | 是                    |

- **分配用户数：** 随机分配到每组的用户数量。
- **转化数：** 每组中通过 Stripe 付费的人数。
- **转化率：** 转化数除以分配用户数。
- **p 值：** 单尾双比例 Z 检验的结果，表明模型 X 更高的转化率是否很可能并非偶然。
- **统计显著？：** p 值是否低于你的 alpha（这里为 0.05）？
- **胜出？：** 如果具有统计显著性，模型 X 就是新的胜出者。
- **已控制第一类错误？：** 我们是否将假阳性风险控制在 alpha 阈值以内？
- **已控制第二类错误？：** 我们的样本量是否提供了足够的检验效能来检测真实效应？

在这次运行中，模型 X 的转化率比对照组高 1 个百分点（2.0% 对比 1.0%），相对提升 100%。p 值 0.012 远低于 0.05，因此我们将其标记为具有统计显著性：模型 X 胜出。由于我们按 75% 的统计检验效能规划了样本量，我们也有信心没有错过真实效应（第二类错误）。而且由于我们将 alpha 设为 0.05，假阳性（第一类错误）的风险也得到了控制。
### 真实案例：HyperWrite 的测试参数
HyperWrite 并没有默认采用教科书式的 95% 置信度和 80% 检验效能。流量很昂贵，追求最大的统计确定性会拖慢学习速度并消耗资金。所选择的 85% 置信度和 60% 检验效能，既能检测到任何实质性的下降（约 30% 的降幅），又避免了对微小差异的过度优化。

测试运行时间越长，转化率往往越高。在这些测试中，一旦达到所需的样本量（N）就停止运行。每个测试组只分配了一小部分新进流量，大部分流量仍保留在经过验证的对照体验上。
### 关于多重比较的说明
测试采用了 A/B/n（“多对一”）设计：每个候选模型（GPT-4.1 和 GPT-4.1-mini）都与生产环境中的对照组（Claude 3.5 Sonnet）进行比较，但候选模型之间不直接比较。

由于上线决策是针对每个变体单独做出的（“如果该组在 α = 0.15 下的单尾非劣效性检验通过就上线，否则放弃”），因此没有进行族错误率校正。对于 k 较小、以对照组为中心的测试，这是标准做法。假阳性风险只适用于实际上线的那一组，而避免 Bonferroni 类型的拆分可以保持检验效能。
### 如何在 Python 中检验 A/B 测试的显著性
为了准确演示 A/B 测试背后的统计原理，下面是一段 10 行的 Python 代码片段，它使用单尾双比例 Z 检验（变体优于对照组）将原始转化计数转换为 p 值。将其粘贴到任意 Python REPL、Colab 或 notebook 中，运行真实实验时换成你自己的数字即可。
```python
# One-tailed two-proportion Z-test
from statsmodels.stats.proportion import proportions_ztest

conversions   = [30, 15]     # [variant, control]
sample_sizes  = [1500, 1500] # [variant, control]

z_stat, p_val = proportions_ztest(
    conversions,
    sample_sizes,
    alternative="larger"      # "larger" → variant > control
)

print(f"Z-statistic = {z_stat:.2f}")
print(f"p-value     = {p_val:.3f}")    # → 0.012 (α = 0.05)
```

如何解读结果：
- 如果 p 值 **≤ 0.05**，说明变体更高的转化率具有统计显著性，可以放心上线，或继续监测以获取更多数据。
- 如果 p 值 **> 0.05**，结果可能只是随机噪声，请收集更多数据，或继续使用对照组。
### 注意事项
- **挑选尾侧 / p 值操纵（p-hacking）：** 在第一个用户进入之前就决定使用单尾还是双尾检验；事后切换会抬高第一类错误（假阳性）。
- **计数过少：** 如果任一组的转化数少于约 10 次，请将 Z 检验换成 Fisher 精确检验或 Wilson/Wald 置信区间。
- **过早偷看：** 在没有进行 α 消耗校正的情况下反复查看数据，会提高假阳性风险。请使用固定样本量或成组序贯设计。
- **用户重叠 / 污染：** 确保同一个用户 ID 不会进入两个组（例如通过退出再登录）。
- **多个挑战者：** 如果你打算从众多变体中挑选唯一“最佳”的一个，请控制族错误率（Bonferroni、Holm）或使用多臂老虎机方法。
- **缓存与提示词漂移：** 确认你的推理层不会把一个模型的响应泄漏到另一个模型的缓存中；在各组之间保持提示词完全一致。

要进一步了解这些陷阱以及如何避免它们，请参阅 Evan Miller 的[《How Not to Run an A/B Test》](https://www.evanmiller.org/how-not-to-run-an-ab-test.html)
### 核心要点
A/B 测试不只适用于落地页或按钮颜色，它对于为你的产品挑选合适的 LLM 也至关重要。将其纳入你的工作流，你就能避免代价高昂的错误，并以用户真正看重的东西为依据发现升级机会：一款值得付费的产品。
## 真实案例：HyperWrite 借助 GPT-4.1 节省成本
模型价格往往会随着能力提升而上涨。HyperWrite 花了几个月时间寻找一个能够匹敌现有模型（Anthropic 的 Claude 3.5 Sonnet）、不损害转化率或用户体验、最好还能降低成本的模型。在几个模型表现更差之后，OpenAI 的 GPT-4.1 交出了一份亮眼的结果：以更低的价格达到了与现有模型相当的 Stripe 转化率。

以下是各变体在 Stripe 转化上的对比：

| 变体                                      | 分配数 | 转化数 | 转化率  | 所需 N | 完成度 | 转化临界值 (≤) | 更差？ |
|----------------------------------------------|---------:|------------:|------:|------:|-------:|-----------------:|:------:|
| anthropic/claude-3.5-sonnet（对照组）        |    4550 |          42 | 0.92% |  3378 |   135% | —                | —      |
| openai/gpt-4.1（变体）                     |    4513 |          58 | 1.29% |  3378 |   134% | 32               | 否     |
| openai/gpt-4.1-mini（变体）                |    4557 |          45 | 0.99% |  3378 |   135% | 33               | 否     |
- **变体：** 模型名称（对照组或挑战者）。
- **分配数：** 随机分配到该组的用户数量。
- **转化数：** 该组中通过 Stripe 付费的用户数。
- **转化率：** 转化数除以分配数。
- **所需 N：** 为非劣效性检验预先计算的样本量目标。
- **完成度：** 分配数除以所需 N（达成目标的进度）。
- **转化临界值 (≤)：** 若该组转化数低于此最大值，则会被标记为“显著差于”对照组。
- **更差？：** 如果该组低于其临界值（即在统计上更差），则为“是”；否则为“否”。

**结果**

- 两个 GPT-4.1 变体都超过了各自的临界值，这意味着两者在统计上都不比对照组差。
- GPT-4.1（完整版）在转化率上与 Claude 3.5 Sonnet 不相上下，同时带来了可观的成本节省。
### 衡量转化需要一些创意和数据
要进行这种分析，你需要一个能够将用户行为与 Stripe 付款事件关联起来的系统。这方面没有通用模板，但 HyperWrite 使用的架构展示了一种实现方式。对于任何用户与 AI 交互并可通过 Stripe 升级的初创公司，都可以调整采用这一工作流。
1. **用户跟踪：** 为每个新注册用户分配一个贯穿其整个生命周期的唯一标识符。
2. **模型分配：** 在注册时将每个用户随机分配到一个测试组（模型变体），并将该分配存储在数据库中。
3. **交互日志：** 记录关键事件（例如首次使用、达到速率限制），以及用户 ID 和模型分配。
4. **转化事件捕获：** 设置 Stripe webhook 来监听 `checkout.session.completed` 事件。事件触发时，将 Stripe 客户与你的内部用户 ID 匹配，并更新数据库以记录付款/转化。
5. **数据汇总：** 定期将测试组分配和转化数据汇总到一张表或一个仪表板中进行分析。
6. **统计检验：** 使用基本的 Z 检验（有许多现成的库/Excel 模板）分析转化率差异是否有意义。

#### 用户工作流
以下是 HyperWrite 上一个用户的使用旅程：

1. **监听 Stripe 事件：** 系统监听 Stripe webhook 发出的 `checkout.session.completed` 事件，该事件在付款成功时触发。

定期检查测试是否已完成：
1. **查询测试组：** 获取分配到每个模型变体的所有用户。

3. **运行统计检验：** 使用单尾双比例 Z 检验（见上一节）检查转化率差异是否具有统计意义。
## 结论与后续步骤

并不总是需要超越现有模型；一个在关键指标上表现“同样好”且成本更低的模型同样很有价值。在本案例中，OpenAI 的 GPT-4.1 在降低成本的同时，达到了与现有模型相当的 Stripe 转化率。

这凸显了将模型评估与 Stripe 驱动的 A/B 测试挂钩的价值：你能获得与收入直接关联的明确答案，而不只是依赖基准或主观印象。

初创公司可以从以下几个方向扩展这类测试：
- **按用户画像或用例细分：** 将受众分组（例如重度用户与新用户、不同行业），看看哪些模型或提示词对每组效果最好。
- **找到收入与成本的最佳平衡点：** 不仅要考虑总收入，还要考虑每个模型的服务成本。最优选择可能是平衡利润，而不仅仅是最大化销售额。
- **监测长期影响：** 不要只看即时转化。跟踪订阅用户生命周期价值、流失率或留存率等指标，以实现可持续增长。
