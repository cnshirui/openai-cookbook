# **GPT Action 库：Google Drive**


## **简介**

本页面为开发者针对特定应用构建 GPT Action 提供说明和指南。在继续之前，请先确保熟悉以下内容：


* [GPT Actions 简介](https://platform.openai.com/docs/actions/introduction)
* [GPT Actions 库简介](https://platform.openai.com/docs/actions/actions-library)
* [从零构建 GPT Action 的示例](https://platform.openai.com/docs/actions/getting-started)

这个 GPT Action 概述了如何连接到 Google 的文件存储系统 **Google Drive**。该 Action 让你可以列出和查询文件名、将文件内容加载到 GPT 中，并最终在 ChatGPT 中将这些数据用作上下文。这组 Action 可以通过 [Google Drive API](https://developers.google.com/drive/api/guides/about-sdk) 中的其他方法进行扩展。如果你想要一个能读取较小文件的通用型 GPT，这会非常合适，例如：


* 会议纪要
* 产品设计文档
* 简短备忘录
* 常见问题解答

如果需要读取更长的内容，例如整本书或包含大量行的复杂 CSV，我们建议构建专门针对 Google Docs 或 Google Sheets 的 GPT。

### 价值与业务示例

用户现在可以利用 ChatGPT 的自然语言能力，直接连接到 Google Drive 中的文件

示例用例：

- 用户需要查找与某个主题相关的文件
- 用户需要为一个深埋在文档中的关键问题找到答案

### **OpenAPI Schema 示例**

创建自定义 GPT 后，将下方文本复制到 Actions 面板中。这里给出了一个示例，展示你可以将哪些内容作为 GPT 的函数。

有疑问？请查看[入门示例](https://platform.openai.com/docs/getting-started)，更详细地了解这一步的工作方式。此外，也可以试试 [ActionsGPT](https://chatgpt.com/g/g-TYEliDU6A-actionsgpt)，这是 OpenAI 为协助构建 Actions 而创建的一个自定义 GPT。三个示例分别是：


* **List Files**：这是列出你云端硬盘中文件的核心 Action。其中包含一些参数，例如 `q`、`includeItemsFromAllDrives,supportsAllDrives`
* **Get Metadata**：当列出文件不奏效时，它可以根据某些结果作为备选方案——例如用户尝试通过“上周的会议”等方式进行搜索时
* **Export**：以字节内容的形式导出。更多信息请参阅 [https://developers.google.com/drive/api/reference/rest/v3/files/export](https://developers.google.com/drive/api/reference/rest/v3/files/export)

一般来说，如果使用 ‘get’，模型会尝试下载文件，这可能不是你想要的。因此建议改用 Export。



```python

{
  "openapi": "3.1.0",
  "info": {
    "title": "Google Drive API",
    "description": "API for interacting with Google Drive",
    "version": "1.0.0"
  },
  "servers": [
    {
      "url": "https://www.googleapis.com/drive/v3"
    }
  ],
  "paths": {
    "/files": {
      "get": {
        "operationId": "ListFiles",
        "summary": "List files",
        "description": "Retrieve a list of files in the user's Google Drive.",
        "parameters": [
          {
            "name": "q",
            "in": "query",
            "description": "Query string for searching files.",
            "required": false,
            "schema": {
              "type": "string"
            }
          },
          {
            "name": "includeItemsFromAllDrives",
            "in": "query",
            "description": "Whether both My Drive and shared drive items should be included in results.",
            "required": false,
            "schema": {
              "type": "string"
            }
          },
          {
            "name": "supportsAllDrives",
            "in": "query",
            "description": "Whether the requesting application supports both My Drives and shared drives.",
            "required": false,
            "schema": {
              "type": "string"
            }
          },
          {
            "name": "pageSize",
            "in": "query",
            "description": "Maximum number of files to return.",
            "required": false,
            "schema": {
              "type": "integer",
              "default": 10
            }
          },
          {
            "name": "pageToken",
            "in": "query",
            "description": "Token for continuing a previous list request.",
            "required": false,
            "schema": {
              "type": "string"
            }
          },
          {
            "name": "fields",
            "in": "query",
            "description": "Comma-separated list of fields to include in the response.",
            "required": false,
            "schema": {
              "type": "string"
            }
          }
        ],
        "responses": {
          "200": {
            "description": "A list of files.",
            "content": {
              "application/json": {
                "schema": {
                  "type": "object",
                  "properties": {
                    "kind": {
                      "type": "string",
                      "example": "drive#fileList"
                    },
                    "nextPageToken": {
                      "type": "string",
                      "description": "Token to retrieve the next page of results."
                    },
                    "files": {
                      "type": "array",
                      "items": {
                        "type": "object",
                        "properties": {
                          "id": {
                            "type": "string"
                          },
                          "name": {
                            "type": "string"
                          },
                          "mimeType": {
                            "type": "string"
                          }
                        }
                      }
                    }
                  }
                }
              }
            }
          }
        }
      }
    },
    "/files/{fileId}": {
      "get": {
        "operationId": "getMetadata",
        "summary": "Get file metadata",
        "description": "Retrieve metadata for a specific file.",
        "parameters": [
          {
            "name": "fileId",
            "in": "path",
            "description": "ID of the file to retrieve.",
            "required": true,
            "schema": {
              "type": "string"
            }
          },
          {
            "name": "fields",
            "in": "query",
            "description": "Comma-separated list of fields to include in the response.",
            "required": false,
            "schema": {
              "type": "string"
            }
          }
        ],
        "responses": {
          "200": {
            "description": "Metadata of the file.",
            "content": {
              "application/json": {
                "schema": {
                  "type": "object",
                  "properties": {
                    "id": {
                      "type": "string"
                    },
                    "name": {
                      "type": "string"
                    },
                    "mimeType": {
                      "type": "string"
                    },
                    "description": {
                      "type": "string"
                    },
                    "createdTime": {
                      "type": "string",
                      "format": "date-time"
                    }
                  }
                }
              }
            }
          }
        }
      }
    },
    "/files/{fileId}/export": {
      "get": {
        "operationId": "export",
        "summary": "Export a file",
        "description": "Export a Google Doc to the requested MIME type.",
        "parameters": [
          {
            "name": "fileId",
            "in": "path",
            "description": "ID of the file to export.",
            "required": true,
            "schema": {
              "type": "string"
            }
          },
          {
            "name": "mimeType",
            "in": "query",
            "description": "The MIME type of the format to export to.",
            "required": true,
            "schema": {
              "type": "string",
              "enum": [
                "application/pdf",
                "application/vnd.openxmlformats-officedocument.wordprocessingml.document",
                "text/plain"
              ]
            }
          }
        ],
        "responses": {
          "200": {
            "description": "The exported file.",
            "content": {
              "application/pdf": {
                "schema": {
                  "type": "string",
                  "format": "binary"
                }
              },
              "application/vnd.openxmlformats-officedocument.wordprocessingml.document": {
                "schema": {
                  "type": "string",
                  "format": "binary"
                }
              },
              "text/plain": {
                "schema": {
                  "type": "string",
                  "format": "binary"
                }
              }
            }
          },
          "400": {
            "description": "Invalid MIME type or file ID."
          },
          "404": {
            "description": "File not found."
          }
        }
      }
    }
  }
}
```

### **在 ChatGPT 中**

在 ChatGPT 中，点击“Authentication”并选择 **“OAuth”**。输入以下信息。

* **Client ID**：使用上述步骤中的 Client ID
* **Client Secret**：使用上述步骤中的 Client Secret
* **Authorization URL**：[https://accounts.google.com/o/oauth2/auth](https://accounts.google.com/o/oauth2/auth)
* **Token URL**：[https://oauth2.googleapis.com/token](https://oauth2.googleapis.com/token)
* **Scope**：[https://www.googleapis.com/auth/drive](https://www.googleapis.com/auth/drive.readonly)
    * **注意**：如需查看更详细的可用 scope 列表，请参阅 [Google 的 OAuth 2.0 指南。](https://developers.google.com/identity/protocols/oauth2/scopes)
* **Token**：Default (POST)
* **Privacy Policy**：[https://policies.google.com/privacy?hl=en-US](https://policies.google.com/privacy?hl=en-US)
