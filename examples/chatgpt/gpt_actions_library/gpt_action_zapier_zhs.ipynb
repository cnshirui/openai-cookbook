# GPT Action 库：Zapier

## 简介

本页面为针对特定应用构建 GPT Action 的开发者提供说明和指南。在继续之前，请先熟悉以下内容： 
- [GPT Actions 简介](https://platform.openai.com/docs/actions)
- [GPT Actions 库简介](https://platform.openai.com/docs/actions/actions-library)
- [从零构建 GPT Action 的示例](https://platform.openai.com/docs/actions/getting-started)

本 GPT Action 概述了如何将 GPT 连接到 **Zapier**。由于大部分配置都在 Zapier 上完成，我们建议阅读这篇 ***[Zapier 关于将 GPT 连接到自定义 Zapier Actions 的实用指南](https://actions.zapier.com/docs/platform/gpt)***。

### 价值 + 业务用例示例

**价值**：用户现在可以将 ChatGPT 中的自定义 GPT 连接到 Zapier，即刻集成技术栈中 6,0000+ 个应用和 20,000+ 个操作。   

**用例示例**： 
- 某组织已经搭建了 Zapier 集成，希望在将其技术生态与 ChatGPT 连接时避免额外的集成工作
- 构建一个日历助手 GPT，用于查询日历事件，并根据参会者的 LinkedIn 资料提供更多背景信息
- 构建一个 CRM GPT，帮助将 Hubspot 连接到 ChatGPT，让销售团队随时随地更新或查看联系人和备注


## 应用信息

### 应用关键链接

开始之前，请先查看该应用的以下链接：
- 应用网站：https://zapier.com
- AI Actions URL：https://actions.zapier.com/gpt/actions/
- 自动 OpenAPI 配置：https://actions.zapier.com/gpt/api/v1/dynamic/openapi.json?tools=meta

### 应用前提条件

开始之前，请确保在 Zapier 中完成以下步骤：
- 通过 [AI Action Manager](https://actions.zapier.com/gpt/actions/) 配置所需的 AI Actions

![zapier_ai_actions.png](../../../images/zapier_ai_actions.png)

![zapier_action_config.png](../../../images/zapier_action_config.png)

### 在 ChatGPT 中

在 ChatGPT 的自定义 GPT 创建界面中，点击“Actions”，然后选择 **“Import from URL”**。输入用于配置 GPT 的 Zapier URL：https://actions.zapier.com/gpt/api/v1/dynamic/openapi.json?tools=meta
