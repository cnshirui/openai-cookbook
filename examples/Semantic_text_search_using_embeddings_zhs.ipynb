## 使用嵌入进行语义文本搜索

通过对搜索查询进行嵌入，然后找出最相似的评论，我们可以以非常高效且低成本的方式对所有评论进行语义搜索。数据集是在 [Get_embeddings_from_dataset Notebook](Get_embeddings_from_dataset.ipynb) 中创建的。

```python
import pandas as pd
import numpy as np
from ast import literal_eval

datafile_path = "data/fine_food_reviews_with_embeddings_1k.csv"

df = pd.read_csv(datafile_path)
df["embedding"] = df.embedding.apply(literal_eval).apply(np.array)
```

这里我们比较查询嵌入与文档嵌入之间的余弦相似度，并展示 top_n 个最佳匹配。

```python
from utils.embeddings_utils import get_embedding, cosine_similarity

# search through the reviews for a specific product
def search_reviews(df, product_description, n=3, pprint=True):
    product_embedding = get_embedding(
        product_description,
        model="text-embedding-3-small"
    )
    df["similarity"] = df.embedding.apply(lambda x: cosine_similarity(x, product_embedding))

    results = (
        df.sort_values("similarity", ascending=False)
        .head(n)
        .combined.str.replace("Title: ", "")
        .str.replace("; Content:", ": ")
    )
    if pprint:
        for r in results:
            print(r[:200])
            print()
    return results


results = search_reviews(df, "delicious beans", n=3)
```

```python
results = search_reviews(df, "whole wheat pasta", n=3)
```

我们可以轻松地搜索这些评论。为了加快计算速度，可以使用一种专门用于加速嵌入搜索的特殊算法。

```python
results = search_reviews(df, "bad delivery", n=1)
```

可以看到，这能立即带来很大的价值。在这个示例中，我们展示了如何快速找到配送失败的案例。

```python
results = search_reviews(df, "spoilt", n=1)
```
